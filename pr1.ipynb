```python
import math

DATA = [
    ['Стабильность', 'Раз в неделю', 'Нет', 'Экосистема;Надежность', 'Технологичность', 'VPN', '21', '80000', '5', 'Да', 'Плохо', 'Да', '3', 'Не важна', 'IT', 'М', 'Нет', '1', 'Экономить и откладывать', 'От одного бренда', '1', 'Андроид'],
    ['Стабильность', 'Каждый день', 'Да', 'Статус;Камера', 'Статус', 'VPN', '21', '90000', '6', 'Да', 'Хорошо', 'Да', '3', 'Важна', 'Продажи', 'М', 'Да', '5', 'Экономить и откладывать', 'От одного бренда', '7', 'Айфон'],
    ['Новшество', 'Раз в неделю', 'Нет', 'Экосистема;Надежность;Мультивыбор', 'Статус', 'VPN', '22', '60', '5', 'Не знаю', 'Хорошо', 'Да', '3', 'Важна', 'Менеджер', 'М', 'Нет', '2', 'Порадовать себя', 'От одного бренда', '4', 'Айфон'],
    ['Стабильность', 'Раз в месяц', 'Нет', 'Надежность', 'Технологичность', 'VPN', '22', '50000', '1', 'Не знаю', 'Плохо', 'Нет', '7', 'Не важна', 'Студент', 'М', 'Нет', '1', 'Экономить и откладывать', 'От разных брендов', '10', 'Айфон'],
    ['Стабильность', 'Раз в неделю', 'Нет', 'Камера;Экосистема;Надежность', 'Технологичность', 'VPN', '22', '30000', '3', 'Не знаю', 'Плохо', 'Нет', '2.5', 'Важна', 'Тех под', 'М', 'Нет', '2', 'Экономить и откладывать', 'От разных брендов', '1', 'Андроид'],
    ['Новшество', 'Раз в неделю', 'Нет', 'Статус;Камера;Экосистема', 'Статус', 'VPN', '22', '70000', '6', 'Да', 'Хорошо', 'Не знаю', '5', 'Важна', 'IT', 'М', 'Нет', '3', 'Экономить и откладывать', 'От одного бренда', '10', 'Айфон'],
    ['Стабильность', 'Раз в месяц', 'Нет', 'Камера', 'Статус', 'VPN', '30', '50000', '7', 'Не знаю', 'Хорошо', 'Не знаю', '4', 'Не важна', 'нету', 'М', 'Нет', '2', 'Порадовать себя', 'От одного бренда', '2', 'Айфон'],
    ['Стабильность', 'Раз в неделю', 'Нет', 'Камера;Надежность', 'Технологичность', 'VPN', '21', '70000', '4', 'Не знаю', 'Хорошо', 'Не знаю', '5', 'Важна', 'инженер ', 'Ж', 'Да', '4', 'Порадовать себя', 'От одного бренда', '6', 'Айфон'],
    ['Стабильность', 'Раз в месяц', 'Нет', 'Надежность', 'Технологичность', 'Запрет', '22', '20000', '8', 'Да', 'Хорошо', 'Не знаю', '6', 'Важна', 'Медиа', 'М', 'Нет', '1', 'Экономить и откладывать', 'От разных брендов', '10', 'Андроид'],
    ['Стабильность', 'Раз в неделю', 'Нет', 'Камера;Надежность', 'Технологичность', 'VPN', '23', '30000', '3', 'Не знаю', 'Хорошо', 'Нет', '5', 'Не важна', 'IT', 'М', 'Да', '2', 'Порадовать себя', 'От разных брендов', '5', 'Андроид'],
    ['Новшество', 'Каждый день', 'Нет', 'Статус;Камера;Экосистема;Надежность', 'Технологичность', 'VPN', '21', '100000', '6', 'Не знаю', 'Плохо', 'Не знаю', '3', 'Не важна', 'IT', 'М', 'Нет', '1', 'Порадовать себя', 'От разных брендов', '1', 'Айфон'],
    ['Стабильность', 'Раз в месяц', 'Нет', 'Статус;Экосистема;Надежность', 'Статус', 'VPN', '22', '70000', '4', 'Да', 'Плохо', 'Нет', '4', 'Не важна', 'Программирование', 'М', 'Нет', '2', 'Порадовать себя', 'От разных брендов', '3', 'Андроид'],
    ['Стабильность', 'Раз в неделю', 'Нет', 'Камера;Надежность', 'Технологичность', 'VPN', '22', '100000', '5', 'Да', 'Хорошо', 'Не знаю', '7', 'Важна', 'Дизайн', 'М', 'Нет', '2', 'Экономить и откладывать', 'От разных брендов', '1', 'Андроид'],
    ['Стабильность', 'Каждый день', 'Нет', 'Камера;Экосистема;Надежность', 'Статус', 'VPN', '22', '40000', '4', 'Не знаю', 'Хорошо', 'Не знаю', '3', 'Важна', 'IT', 'М', 'Нет', '3', 'Экономить и откладывать', 'От одного бренда', '7', 'Айфон'],
    ['Стабильность', 'Каждый день', 'Нет', 'Камера;Экосистема', 'Технологичность', 'VPN', '22', '80000', '1', 'Нет', 'Хорошо', 'Не знаю', '3', 'Важна', 'IT', 'М', 'Нет', '2', 'Порадовать себя', 'От одного бренда', '10', 'Айфон'],
    ['Стабильность', 'Каждый день', 'Да', 'Надежность', 'Технологичность', 'VPN', '22', '40000', '7', 'Да', 'Хорошо', 'Да', '4', 'Важна', 'корпоративные системы', 'М', 'Нет', '0.5', 'Порадовать себя', 'От разных брендов', '8', 'Андроид'],
    ['Стабильность', 'Каждый день', 'Нет', 'Статус;Камера;Экосистема;Надежность', 'Статус', 'VPN', '22', '100.000', '5', 'Да', 'Хорошо', 'Не знаю', '3', 'Важна', 'программист', 'М', 'Нет', '2', 'Экономить и откладывать', 'От одного бренда', '5', 'Айфон'],
    ['Стабильность', 'Каждый день', 'Нет', 'Надежность', 'Технологичность', 'VPN', '22', '25000', '7', 'Да', 'Плохо', 'Да', '5', 'Важна', 'Студент', 'М', 'Нет', '1', 'Экономить и откладывать', 'От разных брендов', '5', 'Андроид'],
    ['Новшество', 'Каждый день', 'Да', 'Статус', 'Статус', 'VPN', '34', '100000', '10', 'Да', 'Хорошо', 'Не знаю', '3', 'Не важна', 'Бизнес ', 'Ж', 'Да', '3', 'Порадовать себя', 'От одного бренда', '1', 'Андроид'],
    ['Новшество', 'Каждый день', 'Нет', 'Экосистема;Надежность', 'Технологичность', 'VPN', '21', '70000', '9', 'Да', 'Хорошо', 'Да', '4', 'Важна', 'IT', 'М', 'Нет', '1', 'Экономить и откладывать', 'От разных брендов', '5', 'Андроид'],
    ['Стабильность', 'Раз в месяц', 'Нет', 'Надежность', 'Технологичность', 'VPN', '21', '30000', '3', 'Да', 'Плохо', 'Нет', '5', 'Не важна', 'Студент ', 'М', 'Нет', '1', 'Экономить и откладывать', 'От разных брендов', '8', 'Андроид'],
    ['Стабильность', 'Раз в неделю', 'Да', 'Камера;Экосистема;Надежность', 'Технологичность', 'VPN', '21', '90.000', '5', 'Да', 'Хорошо', 'Да', '3', 'Важна', 'Студент', 'М', 'Да', '4', 'Порадовать себя', 'От разных брендов', '3', 'Айфон'],
    ['Стабильность', 'Раз в неделю', 'Нет', 'Экосистема;Надежность', 'Технологичность', 'VPN', '22', '70000', '7', 'Да', 'Плохо', 'Да', '5', 'Не важна', 'ИТ', 'М', 'Нет', '1', 'Экономить и откладывать', 'От разных брендов', '8', 'Андроид'],
    ['Новшество', 'Каждый день', 'Да', 'Камера', 'Технологичность', 'VPN', '22', '50000', '10', 'Да', 'Хорошо', 'Да', '2', 'Важна', 'Дизайн', 'Ж', 'Да', '2', 'Экономить и откладывать', 'От одного бренда', '8', 'Андроид'],
    ['Стабильность', 'Раз в неделю', 'Нет', 'Экосистема;Надежность', 'Технологичность', 'VPN', '22', '20000', '5', 'Да', 'Плохо', 'Нет', '4 ', 'Не важна', 'Программист ', 'М', 'Нет', '1', 'Экономить и откладывать', 'От разных брендов', '5', 'Андроид'],
    ['Новшество', 'Раз в месяц', 'Нет', 'Экосистема;Надежность', 'Технологичность', 'VPN', '22', '60000', '5', 'Не знаю', 'Хорошо', 'Да', '6', 'Важна', 'Ит', 'М', 'Нет', '1', 'Экономить и откладывать', 'От разных брендов', '3', 'Андроид'],
    ['Стабильность', 'Раз в месяц', 'Нет', 'Экосистема;Надежность', 'Технологичность', 'Запрет', '22', '40000', '5', 'Да', 'Хорошо', 'Да', '7', 'Важна', 'IT', 'М', 'Нет', '1', 'Экономить и откладывать', 'От одного бренда', '5', 'Андроид'],
    ['Новшество', 'Раз в неделю', 'Нет', 'Надежность', 'Технологичность', 'VPN', '21', '70', '5', 'Нет', 'Хорошо', 'Не знаю', '3', 'Не важна', 'ИТ', 'М', 'Нет', '1', 'Порадовать себя', 'От одного бренда', '2', 'Айфон'],
    ['Новшество', 'Раз в месяц', 'Нет', 'Надежность', 'Технологичность', 'Запрет', '21', '65000', '6', 'Да', 'Хорошо', 'Да', '3', 'Не важна', 'ИТ', 'М', 'Нет', '2', 'Порадовать себя', 'От разных брендов', '10', 'Андроид'],
    ['Стабильность', 'Раз в неделю', 'Нет', 'Статус;Камера;Экосистема;Надежность', 'Статус', 'VPN', '23', '100000', '4', 'Да', 'Хорошо', 'Да', '4', 'Не важна', 'Разработка ', 'М', 'Да', '2', 'Порадовать себя', 'От одного бренда', '3', 'Айфон'],
    ['Стабильность', 'Каждый день', 'Да', 'Статус;Камера;Экосистема;Надежность', 'Статус', 'VPN', '21', '60', '4', 'Да', 'Хорошо', 'Нет', '4', 'Важна', 'предприниматель ', 'М', 'Да', '2', 'Порадовать себя', 'От разных брендов', '1', 'Айфон'],
    ['Новшество', 'Каждый день', 'Да', 'Статус;Камера;Надежность', 'Статус', 'VPN', '20', '120', '6', 'Да', 'Хорошо', 'Не знаю', '3', 'Не важна', 'ИТ', 'Ж', 'Да', '1', 'Порадовать себя', 'От одного бренда', '5', 'Айфон'],
    ['Стабильность', 'Раз в неделю', 'Нет', 'Камера;Экосистема', 'Технологичность', 'VPN', '22', '60000', '4', 'Да', 'Хорошо', 'Да', '3', 'Важна', 'It', 'М', 'Нет', '1', 'Порадовать себя', 'От разных брендов', '4', 'Андроид'],
    ['Стабильность', 'Каждый день', 'Нет', 'Камера', 'Технологичность', 'VPN', '20', '50000', '6', 'Да', 'Хорошо', 'Да', '7', 'Не важна', 'It', 'Ж', 'Да', '3', 'Экономить и откладывать', 'От разных брендов', '5', 'Андроид'],
    ['Стабильность', 'Раз в неделю', 'Нет', 'Надежность', 'Технологичность', 'VPN', '22', '50000', '8', 'Да', 'Хорошо', 'Да', '5', 'Важна', 'IT', 'М', 'Нет', '1', 'Экономить и откладывать', 'От разных брендов', '3', 'Андроид'],
    ['Стабильность', 'Каждый день', 'Нет', 'Экосистема;Надежность', 'Технологичность', 'VPN', '22', '60000', '2', 'Да', 'Хорошо', 'Нет', '4', 'Важна', 'Программирование', 'М', 'Нет', '1', 'Экономить и откладывать', 'От одного бренда', '5', 'Айфон'],
    ['Стабильность', 'Раз в неделю', 'Нет', 'Экосистема;Надежность', 'Технологичность', 'VPN', '21', '25000', '5', 'Да', 'Плохо', 'Да', '4', 'Важна', 'IT', 'М', 'Нет', '1', 'Экономить и откладывать', 'От разных брендов', '4', 'Андроид'],
    ['Новшество', 'Раз в неделю', 'Да', 'Камера;Надежность', 'Технологичность', 'VPN', '22', '50000', '5', 'Да', 'Плохо', 'Не знаю', '2', 'Важна', 'Крутой', 'М', 'Нет', '2', 'Порадовать себя', 'От разных брендов', '10', 'Айфон'],
    ['Новшество', 'Раз в месяц', 'Нет', 'Статус;Экосистема;Надежность', 'Технологичность', 'Запрет', '22', '100000', '4', 'Да', 'Хорошо', 'Нет', '4', 'Не важна', 'бар', 'М', 'Нет', '1', 'Экономить и откладывать', 'От одного бренда', '3', 'Айфон'],
    ['Новшество', 'Каждый день', 'Нет', 'Статус;Камера;Экосистема;Надежность', 'Технологичность', 'VPN', '21', '100000', '7', 'Да', 'Хорошо', 'Нет', '1.5', 'Не важна', 'Спорт', 'М', 'Да', '2', 'Экономить и откладывать', 'От одного бренда', '10', 'Айфон'],
    ['Новшество', 'Каждый день', 'Да', 'Экосистема', 'Статус', 'VPN', '5', '3', '6', 'Нет', 'Хорошо', 'Нет', '1', 'Важна', '5', 'М', 'Нет', '5', 'Экономить и откладывать', 'От разных брендов', '2', 'Андроид'],
]

def parse_budget(raw):
    s = raw.replace(" ", "").replace(",", ".").replace(".", "")
    value = float(s)
    if value < 1000:
        value *= 1000
    return value

def sphere_category(raw):
    s = raw.strip().lower()
    if "студент" in s:
        return "Студент"
    if any(k in s for k in ("it", "ит", "програм", "разраб", "инжен", "тех")):
        return "IT"
    if "дизайн" in s or "медиа" in s:
        return "Дизайн и медиа"
    if any(k in s for k in ("продаж", "бизнес", "менедж", "предпри")):
        return "Бизнес"
    return "Другое"

def encode_row(row):
    v = []

    v.append(1.0 if "Стабильность" in row[0] else 0.0)

    photo = {"Каждый день": 3.0, "Раз в неделю": 2.0, "Раз в месяц": 1.0}
    v.append(photo.get(row[1]))

    v.append(1.0 if row[2] == "Да" else 0.0)

    pay = {"Экосистема": 1.0, "Надежность": 2.0, "Камера": 3.0, "Статус": 4.0}
    for option in pay:
        v.append(pay[option] if option in row[3] else 0.0)

    v.append(1.0 if "Статус" in row[4] else 0.0)

    v.append(1.0 if "VPN" in row[5] else 0.0)

    v.append(float(row[6].replace(",", ".")))

    v.append(parse_budget(row[7]))

    v.append(float(row[8].replace(",", ".")))

    v.append({"Да": 1.0, "Не знаю": 0.5, "Нет": 0.0}.get(row[9]))

    v.append(1.0 if row[10] == "Хорошо" else 0.0)

    v.append({"Да": 1.0, "Не знаю": 0.5, "Нет": 0.0}.get(row[11]))

    v.append(float(row[12].strip().replace(",", ".")))

    v.append(1.0 if "Важна" in row[13] else 0.0)

    sphere = {
        "IT": 1.0,
        "Студент": 2.0,
        "Дизайн и медиа": 3.0,
        "Бизнес": 4.0,
        "Другое": 5.0,
    }

    v.append(sphere.get(sphere_category(row[14])))

    v.append(1.0 if row[15].strip().upper().startswith("М") else 0.0)

    v.append(1.0 if row[16] == "Да" else 0.0)

    v.append(float(row[17].replace(",", ".")))

    v.append(1.0 if "Экономить" in row[18] else 0.0)

    v.append(1.0 if "одного" in row[19] else 0.0)

    v.append(float(row[20]))

    return v

def fit_minmax(matrix):
    n_feat = len(matrix[0])

    mins = []
    for j in range (n_feat):
        column_values = []
        for row in matrix:
            column_values.append(row[j])
        mins.append(min(column_values))

    maxs = []
    for j in range (n_feat):
        column_values = []
        for row in matrix:
            column_values.append(row[j])
        maxs.append(max(column_values))

    return mins, maxs

def scale(vector, mins, maxs):
    out = []

    for i in range (len(vector)):
        x = vector[i]
        lowest = mins[i]
        highest = maxs[i]
        difference = highest - lowest
        out.append(0.0 if difference == 0 else (x - lowest) / difference)

    return out

def euclidean(a, b):

    return math.sqrt(sum((a[i] - b[i]) ** 2 for i in range(len(a))))

def count_votes(neighbors):
    votes = {}
    for pair in neighbors:
        brand = pair[1]
        if brand in votes:
            votes[brand] += 1
        else:
            votes[brand] = 1

    return votes

def predict_knn(new_vector, features, brand, k=4, show_ties=True):
    distances_and_brand = []
    for i in range(len(features)):
        distance = euclidean(new_vector, features[i])
        distances_and_brand.append((distance, brand[i]))

    distances_and_brand.sort()

    neighbors = []
    for i in range(k):
        neighbors.append(distances_and_brand[i])

    while True:
        votes = count_votes(neighbors)
        if len(votes) == 1 or len(neighbors) == 1:
            break

        labels = list(votes.keys())
        if votes[labels[0]] != votes[labels[1]]:
            break

        if show_ties:
            print(f"  ничья {votes} => убираем дальнего (расстояние = {neighbors[-1][0]:.3f}, {neighbors[-1][1]})")

        neighbors.pop()

    if votes.get("Айфон", 0) >= votes.get("Андроид", 0):
        winner = "Айфон"
    else:
        winner = "Андроид"

    confidence = votes[winner] / len(neighbors) * 100
    return winner, confidence, neighbors

def check_different_k():
    k_values = list(range(2, 21))

    print("\nПроверка разных значений k:")
    print("-" * 47)
    print(f"{'k':<5} {'Верно':<10} {'Ошибок':<10} {'Accuracy':<10}")
    print("-" * 47)

    for k in k_values:
        correct = 0

        for test_index in range(len(DATA)):
            test_row = DATA[test_index]
            test_answers = test_row[:-1]
            real_brand = test_row[-1]

            train_data = []
            for i in range(len(DATA)):
                if i != test_index:
                    train_data.append(DATA[i])

            train_features = []
            train_brands = []
            for row in train_data:
                train_features.append(row[:-1])
                train_brands.append(row[-1])

            encoded_train = []
            for row in train_features:
                encoded_train.append(encode_row(row))

            mins, maxs = fit_minmax(encoded_train)

            X_train = []
            for vector in encoded_train:
                X_train.append(scale(vector, mins, maxs))

            encoded_test = encode_row(test_answers)
            X_test = scale(encoded_test, mins, maxs)

            predicted_brand, confidence, neighbors = predict_knn(
                X_test,
                X_train,
                train_brands,
                k=k,
                show_ties=False
            )

            if predicted_brand == real_brand:
                correct += 1

        total = len(DATA)
        errors = total - correct
        accuracy = correct / total * 100

        print(f"{k:<5} {correct:<10} {errors:<10} {accuracy:>7.1f}%")

def main():

    features = [row[:-1] for row in DATA]
    brands = [row[-1] for row in DATA]

    android = 0
    iphone = 0
    for brr in brands:
        if brr == "Андроид":
            android += 1
        if brr == "Айфон":
            iphone += 1

    print(f"\nАнкет: {len(brands)} (Андроид={android}, Айфон={iphone})")

    encoded = []
    for r in features:
        encoded.append(encode_row(r))

    mins, maxs = fit_minmax(encoded)

    X = []
    for v in encoded:
        X.append(scale(v, mins, maxs))

    new_answers = [
        "Стабильность",
        "Каждый день",
        "Нет",
        "Камера;Экосистема",
        "Технологичность",
        "VPN",
        "21",
        "20000",
        "7",
        "Да",
        "Хорошо",
        "Да",
        "2",
        "Важна",
        "IT",
        "М",
        "Да",
        "2",
        "Экономить и откладывать",
        "От одного бренда",
        "8",
    ]

    test_new = scale(encode_row(new_answers), mins, maxs)

    print("\nПредсказание для тестовой анкеты:")
    brand_new_answer, confidence_new_answer, neighbors_new_answer = predict_knn(test_new, X, brands, k=4)

    print(f"    скорее всего {brand_new_answer} (уверенность {confidence_new_answer:.0f}%)")

    print("\nCоседи:")
    for result_distance, result_brand in neighbors_new_answer:
        print(f"    {result_distance:.3f}  {result_brand}")

    check_different_k()

if __name__ == "__main__":
    main()
```
